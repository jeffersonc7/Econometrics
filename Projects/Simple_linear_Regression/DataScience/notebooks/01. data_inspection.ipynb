# __Data Inspection__

## Dictionary

```python
# Libraries
import pandas as pd 
from huggingface_hub import hf_hub_download
```

```python
# Dictionary for paths (dictionary)

paths_dict = {
    'path_id': 'Jefferson7c/Econometrics',
    'file_name': ['Diccionario de Datos_persona_tics_2025_07.ODS',
                  'Diccionario de Datos_vivienda_hogar_tics_2025_07.ODS'],
    'repotype': 'dataset'
}
```

```python
# Dictionaries import

# ENEMDU person ICT
dictionary_1 = pd.read_excel(
    hf_hub_download(paths_dict['path_id'],
                    paths_dict['file_name'][0],
                      repo_type=paths_dict['repotype']),
                        engine='odf'
)

# Dwelling Household ICT
dictionary_2 = pd.read_excel(
    hf_hub_download(paths_dict['path_id'], 
                    paths_dict['file_name'][1],
                      repo_type=paths_dict['repotype']),
                        engine='odf'
)

display(dictionary_1, dictionary_2)
```

## Data

```python
# Dictionary for paths (data)

path = {
    'path_id': 'Jefferson7c/Econometrics',
    'file_name': ['enemdu_vivienda_hogar_tics_2025_07.csv', 'enemdu_persona_tics_2025_07.csv'],
    'repotype': 'dataset'
}
```

```python
# Data import

# ENEMDU Dwelling Household ICT
data1 = pd.read_csv(hf_hub_download(path['path_id'],
                                     path['file_name'][0],
                                       repo_type=path['repotype']),
                                         sep=';', low_memory=False)

# ENEMDU person ICT
data2 = pd.read_csv(hf_hub_download(path['path_id'],
                                     path['file_name'][1],
                                       repo_type=path['repotype']),
                                         sep=';', low_memory=False)

display(data1.head(4),data2.head(4))
```

```python
data1.info(verbose=True)
```

## Variables of Interest

```python
# Identifier Variables

data_id = data2[
    ['id_persona', 'id_hogar', 'id_vivienda',
     'ciudad', 'area', 'conglomerado', 'panelm',
     'vivienda', 'hogar', 'p01', 'periodo']
].copy()

display(data_id.info(), data_id)
```

```python
# Sampling Design

data_survey= data2[['fexp', 'estrato', 'upm']].copy()
display(data_survey.info(), data_survey)
```

```python
# Internet Hours & Labor Income

data2_1= data2[['pt3b','ingrl']].copy()
display(data2_1.info(),data2_1)
```

```python
# Data Describe
data2_1.describe()
```

## Import Data

```python
data_final = pd.concat([data_id,data_survey,data2_1],axis=1)
data_final
```

```python
# Store identifiers as strings to preserve every digit

data_export = data_final.astype({
    'id_persona': 'string',
    'id_hogar': 'string',
    'id_vivienda': 'string'
})
data_export
```

```python
from pathlib import Path

# Save the inspection dataset locally
file_path = Path('../data/interim/data_inspection.parquet')
data_export.to_parquet(file_path, index=False)
```
