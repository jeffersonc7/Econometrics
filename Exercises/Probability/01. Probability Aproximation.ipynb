## __Exercise 2.16 — Probability Approximation__

Sea una variable aleatoria con distribución:

$$Y \sim N(5,100)$$

Se desea calcular:

$$P(Y < 3.6)$$

No se dispone de un libro de texto ni de una tabla de probabilidades de la distribución normal, como la Tabla 1 del Apéndice.

Sin embargo, se cuenta con un ordenador y un programa capaz de generar observaciones independientes e idénticamente distribuidas (*i.i.d.*) de $N(5,100)$.

**Pregunta:** ¿Cómo utilizarías el ordenador para obtener una aproximación precisa de $P(Y < 3.6)$?


```python
# Convergence Mean

import numpy as np 

# fix seed
rng = np.random.default_rng(seed=10)

# Parameters
mu = 5
sigma = np.sqrt(100)

#  i.i.d observations
sample_size = np.logspace(0,7, num=50,dtype=int)
sample_means = []

# Sample means (Montecarlo Simulation)
for i in sample_size: 
    sample = rng.normal(loc=mu, scale=sigma, size=i)
    sample_means.append(float(sample.mean()))

# First 5 sample means
sample_means[:5]
```

```python
# Convergence Graph

import seaborn as sns
import matplotlib.pyplot as plt 

# Figure
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

# Plot
sns.lineplot(x=sample_size, y=sample_means, marker="o")

# Plot Settings
plt.axhline(y=mu, color="red", linestyle="--")
plt.xscale("log")
plt.xlabel("Sample Size")
plt.ylabel("Sample Mean")
plt.title("Sample Mean Convergence")
plt.show()
```

```python
# Probability 

probability = np.mean(sample < 3.6)
print(f'Probability : {probability}')
```

```python
# Probability Convergence

probabilities = []

# Montecarlo Simulation
for n in sample_size:
    sample = rng.normal(loc=mu,scale=sigma,size=n)
    probabilities.append(float(np.mean(sample<3.6)))

probabilities[:5]
```

```python
# Convergence Graph

import seaborn as sns
import matplotlib.pyplot as plt 

# Figure
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

# Plot
sns.lineplot(x=sample_size, y=probabilities, marker="o")

# Plot Settings
plt.axhline(y=probability, color="red", linestyle="--")
plt.xscale("log")
plt.xlabel("Sample Size")
plt.ylabel("Sample Probability")
plt.title("Sample Probability Convergence")
plt.show()
```
